```python
# ==========================================================
# IEEE AI Track - Pre-Session 0 Warm-Up Task
# Student Name: [Mariam Ashraf]
# ==========================================================

# ----------------------------------------------------------
# Task 1: Create Your Student Profile Dictionary
# ----------------------------------------------------------
# Create a dictionary called student_profile containing 3 keys:
#   1. "name"          -> Your full name (string)
#   2. "faculty_major" -> Your college / department (string)
#   3. "ai_interest"   -> The AI topic you're excited to learn (string)
#
# Hint: In Python, dictionaries use curly braces with key:value pairs.
# Example format: my_dict = {"key1": "value1", "key2": "value2"}

# TODO: Define the dictionary below
student_profile = {
"name" :"mariam",
"faculty_major" :"AI",
"ai_interest": "AI in cybersecurity, computer vision"
}


# ----------------------------------------------------------
# Task 2: Build the Text Analyzer Function
# ----------------------------------------------------------
# Write a function named analyze_bio(bio_text) that:
#   1. Calculates the total number of characters (including spaces).
#   2. Calculates the total number of words.
#   3. Returns both values (character_count, word_count).
#
# Hints:
#   - To find the length of a string, use Python's built-in len(...) function.
#   - To break a sentence into a list of words separated by spaces, use .split().
#   - To find how many items are in that word list, pass it to len(...).
#   - To return multiple values from a function, separate them with a comma:
#     return val1, val2

def analyze_bio(bio_text):
    char_count = len(bio_text)
    word_list = bio_text.split()
    word_count = len(word_list)
    return char_count, word_count


# ----------------------------------------------------------
# Task 3: Test and Print Your Results
# ----------------------------------------------------------
# 1. Write a 1-2 sentence bio describing your goals in AI.
# 2. Call your analyze_bio() function on that string.
# 3. Print your profile data and the analysis using f-strings.
#
# Hint: An f-string lets you insert variables directly into text:
# Example: print(f"Hello, my name is {my_name}")

my_bio = ("Hello, I am a third-year student who is passionate about AI. "
          "AI is now a part of our everyday life, "
          "and developing yourself in it is now a required skill.")

# TODO: Call your function and unpack the results
total_chars, total_words = analyze_bio(my_bio)

# TODO: Print a clean formatted summary showing:
#   - Your name, major, and AI interest from the student_profile dictionary
#   - Your bio text
#   - The character count and word count
print(f"My name is {student_profile['name']}, Major: {student_profile['faculty_major']} and AI interest: {student_profile['ai_interest']}")
print("=" * 40)
print("My Bio:")
print(my_bio)
print("=" * 40)
print(f"Character Count: {total_chars}")
print(f"Word Count: {total_words}")
print("=" * 40)
```
